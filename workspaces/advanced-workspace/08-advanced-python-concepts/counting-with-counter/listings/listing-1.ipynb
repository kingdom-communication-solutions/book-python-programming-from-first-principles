From *Python Programming from First Principles*, Advanced Python. [Read this section](/books/001-python-programming-from-first-principles/counting-with-counter.html).

Predict the output, then run the listing.

````python
from collections import Counter

text = "to be or not to be that is the question"
counts = Counter(text.split())

print(counts)
print(counts.most_common(2))
````
