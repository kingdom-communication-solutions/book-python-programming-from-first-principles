# Iteration Tools with `itertools`: listing 1

From *Python Programming from First Principles*, Advanced Python. [Read this section](/books/001-python-programming-from-first-principles/iteration-tools-itertools.html).

Predict the output, then run the listing.

````python
from itertools import combinations

students = ["Ada", "Grace", "Alan"]

for pair in combinations(students, 2):
    print(pair)
````
