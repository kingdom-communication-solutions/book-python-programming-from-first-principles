Predict the output, then run the listing.

```python
%%writefile run_suite.py
# run_suite.py
import unittest
from test_math_helpers import TestMathHelpers
from test_strings import TestStringMethods

loader = unittest.defaultTestLoader
suite = unittest.TestSuite([
    loader.loadTestsFromTestCase(TestMathHelpers),
    loader.loadTestsFromTestCase(TestStringMethods),
])

runner = unittest.TextTestRunner()
runner.run(suite)
```
