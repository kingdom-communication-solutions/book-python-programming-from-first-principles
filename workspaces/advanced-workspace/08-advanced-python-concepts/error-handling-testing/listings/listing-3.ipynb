Predict the output, then run the listing.

````python
try:
    n = 0
    res = 100 / n
except ZeroDivisionError:
    print("You can't divide by zero!")
except ValueError:
    print("Invalid value provided.")
else:
    print("Result:", res)
finally:
    print("Execution complete.")
````
