# Priority Queues with `heapq`: listing 1

From *Python Programming from First Principles*, Advanced Python. [Read this section](/books/001-python-programming-from-first-principles/priority-queues-with-heapq.html).

Predict the output, then run the listing.

````python
import heapq

# (priority, description)
tasks = [
    (3, "Update documentation"),
    (1, "Fix production bug"),
    (2, "Reply to customer emails"),
]

heapq.heapify(tasks)  # turn list into a heap

while tasks:
    priority, description = heapq.heappop(tasks)
    print(f"Handling (priority {priority}): {description}")
````

```text
Handling (priority 1): Fix production bug
Handling (priority 2): Reply to customer emails
Handling (priority 3): Update documentation
```
