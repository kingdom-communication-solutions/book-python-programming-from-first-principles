# Dictionaries: listing 13

From *Python Programming from First Principles*, Python Basics. [Read this section](/books/001-python-programming-from-first-principles/collection-dictionaries.html).

Run the first cell once. It repeats the earlier code of this section that the listing builds on, so it may print earlier results.

````python
# Earlier code from this section

student = {
    "name": "Alice",
    "age": 22,
    "major": "Computer Science"
}

print(student)

print(student["name"])
print(student.get("age"))

student["age"] = 23
student["major"] = "Data Science"
print(student)

student.update({"age": 24, "year": "Final"})
print(student)

student["GPA"] = 3.9
print(student)

student.pop("year")       # removes item by key
print(student)

student.popitem()         # removes the last inserted key–value pair
print(student)

del student["age"]        # removes key entirely
print(student)

student.clear()           # removes all items
print(student)

student = {"name": "Alice", "age": 24, "major": "Data Science"}

# Loop through keys
for key in student:
    print(key)

# Loop through values
for value in student.values():
    print(value)

# Loop through both keys and values
for key, value in student.items():
    print(key, ":", value)

student = {"name": "Alice", "age": 24}
copy_student = student.copy()
copy_student["name"] = "Eve"

print(student)
print(copy_student)

students = {
    "A1": {"name": "Alice", "age": 22},
    "B2": {"name": "Bob", "age": 23},
    "C3": {"name": "Charlie", "age": 21}
}

print(students["A1"]["name"])
````

Predict the output, then run the listing.

````python
for student_id, info in students.items():
    print(student_id, "=>", info["name"], "-", info["age"])
````
