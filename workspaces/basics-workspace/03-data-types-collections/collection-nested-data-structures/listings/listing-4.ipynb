````python
# Earlier code from this section

books = [
    {"title": "To Kill a Mockingbird", "author": "Harper Lee", "year": 1960},
    {"title": "1984", "author": "George Orwell", "year": 1949},
    {"title": "The Great Gatsby", "author": "F. Scott Fitzgerald", "year": 1925}
]

print(books[0]["title"])  # Access the first book's title
print(books[1]["author"]) # Access the second book's author

board = [
    ["X", "O", ""],
    ["", "X", ""],
    ["O", "", ""],
]

print(board[1][1])   # the center cell
board[2][2] = "X"    # place an X in the bottom-right cell
print(board[2])

store_sections = {
    "Fruits": ["Apples", "Bananas", "Mangoes"],
    "Vegetables": ["Carrots", "Broccoli", "Spinach"],
    "Bakery": ["Bread", "Croissant"]
}

# Add a new item
store_sections["Bakery"].append("Muffins")
print(store_sections["Bakery"])
````

Predict the output, then run the listing.

````python
customers = {
    "C001": {"name": "Alice", "email": "alice@example.com", "points": 1200},
    "C002": {"name": "Bob", "email": "bob@example.com", "points": 800},
}

print(customers["C001"]["name"])  # Access Alice’s name
customers["C002"]["points"] += 200  # Update Bob’s points
print(customers["C002"])
````
