````python
# Earlier code from this section

fruits = {"apple", "banana", "cherry"}
print(fruits)

empty_set = set()
print(type(empty_set))

visits = ["Paris", "Rome", "Paris", "Lima", "Rome"]

print(len(visits))
print(len(set(visits)))

fruits = {"apple", "banana", "cherry"}

print("banana" in fruits)
print("orange" in fruits)

fruits = {"apple", "banana"}

fruits.add("cherry")
print(fruits)

fruits.update(["mango", "orange"])
print(fruits)

fruits = {"apple", "banana", "cherry"}

fruits.remove("banana")
print(fruits)

fruits.discard("pear")  # no error if missing
print(fruits)

removed = fruits.pop()
print("Removed:", removed)
print("Remaining:", fruits)

fruits.clear()
print("Cleared set:", fruits)

colors = {"red", "green", "blue"}

for color in colors:
    print(color)

a = {1, 2, 3}
b = {3, 4, 5}

print(a.union(b))
print(a | b)

print(a.intersection(b))
print(a & b)

print(a.difference(b))
print(a - b)
print(b - a)
````
