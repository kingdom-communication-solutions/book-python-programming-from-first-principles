### Background

Letters are one of the central sources for historians. They show who knew whom, who wrote from where and when news traveled. Digital humanities projects such as *Mapping the Republic of Letters* at Stanford University collect the metadata of thousands of letters written by early modern scholars (their dates, places, writers and recipients) to study networks of correspondence [@rofl; @edelstein2017]. Before such questions can be asked, the data must be extracted from the transcriptions, and dates in particular are written in many forms: *14 March 1851*, *March 14th, 1851* or *14.3.51*. Regular expressions are a common tool for this first step.

In this project, you work with six short letters written for this book in the style of the mid-nineteenth century; the people and events are invented. As in a real archive, each transcription starts with a **shelfmark**, the catalogue number under which the archive stores the letter, such as `MS 2047`.
