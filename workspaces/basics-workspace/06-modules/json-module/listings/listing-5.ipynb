```python
# Earlier code from this section

import json

data = '{"name": "Alice", "age": 25, "city": "Paris"}'
person = json.loads(data)
print(person["city"])

student = {
    "name": "Bob",
    "age": 22,
    "courses": ["Math", "AI"]
}

json_string = json.dumps(student)
print(json_string)

print(json.dumps(student, indent=4, sort_keys=True))

print(json.loads(json.dumps({"point": (2, 3)})))
print(json.dumps({1: "one"}))
```
