### Background

People have recorded the weather with instruments since the seventeenth century. The **Central England Temperature** series is the longest instrumental temperature record in the world: its monthly values begin in 1659, reconstructed by the climatologist Gordon Manley from historical observations [@manley1974], and its daily values begin in 1772 [@parker1992]. Today the Met Office Hadley Centre maintains the series and publishes it as open data. It does not come from a single thermometer; it combines several stations that represent a roughly triangular area of central England.

Climatologists summarize such readings with measures like the mean and the median, and they compare the climate of different periods over 30 years at a time, the **climate normals** defined by the World Meteorological Organization [@wmo2017]. A single month cannot show a change in climate, but it is enough to practice the calculations. In this project, you summarize the daily mean temperatures of October 1825 and October 2025.
