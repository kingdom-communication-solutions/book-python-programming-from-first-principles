# The `try` and `except` Blocks: listing 7

From *Python Programming from First Principles*, Python Basics. [Read this section](/books/001-python-programming-from-first-principles/try-except.html).

Run the first cell once. It repeats the earlier code of this section that the listing builds on, so it may print earlier results.

````python
# Earlier code from this section

print(int("42") + 1)

text = "forty-two"

try:
    number = int(text)
    print("Converted:", number)
except ValueError:
    print("Please enter digits, such as 42.")

prices = {"tea": 2.5, "coffee": 3.0}

for item in ["tea", "juice"]:
    try:
        print(item, prices[item])
    except KeyError:
        print(item, "is not on the menu")

def average(total, count):
    try:
        return total / count
    except ZeroDivisionError:
        return None

print(average(12, 4))
print(average(12, 0))

try:
    age = int("abc")
except ValueError as error:
    print("Could not convert:", error)
````

Predict the output, then run the listing.

````python
def check_age(age):
    if age < 0:
        raise ValueError("age cannot be negative")
    return age

print(check_age(30))

try:
    check_age(-5)
except ValueError as error:
    print("Rejected:", error)
````
