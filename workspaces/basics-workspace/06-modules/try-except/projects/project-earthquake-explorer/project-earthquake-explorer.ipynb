### Background

In 1935, the seismologist Charles Richter introduced a scale that measures the size of an earthquake by the largest movement recorded on a standard seismograph [@richter1935]. Seismologists today mostly use the **moment magnitude scale**, introduced by Thomas Hanks and Hiroo Kanamori in 1979, which is based on the physical size of the fault rupture [@hanks1979]. Both scales are **logarithmic**: one unit more means about 32 times more energy released, so a magnitude 7 earthquake releases about a thousand times the energy of a magnitude 5.

Agencies such as the U.S. Geological Survey record earthquakes worldwide and publish the records as open data. In this project, you explore a saved copy of the 722 earthquakes of magnitude 4.5 and above in the 30 days before 3 November 2026: the strongest and most recent, how many fall into each magnitude band and, as an extension, which were nearest to a city. Downloads and saved files can fail, so the program must also cope with a missing or damaged file.
