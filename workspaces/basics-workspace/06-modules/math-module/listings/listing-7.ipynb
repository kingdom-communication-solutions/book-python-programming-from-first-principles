````python
# Earlier code from this section

import math

print("Pi constant:", math.pi)
print("Euler’s number:", math.e)

radius = 5
circumference = 2 * math.pi * radius
print("Circumference of the circle:", circumference)

print("Square root of 25:", math.sqrt(25))
print("2 raised to the power of 5:", math.pow(2, 5))
print("Exponential of 2:", math.exp(2))

print("Natural log of e:", math.log(math.e))
print("Log base 10 of 1000:", math.log10(1000))
print("Log base 2 of 8:", math.log2(8))

print("Floor of 3.7:", math.floor(3.7))   # Rounds down
print("Ceil of 3.2:", math.ceil(3.2))     # Rounds up
print("Absolute value of -5.3:", math.fabs(-5.3))
````

Predict the output, then run the listing.

````python
angle = math.pi / 4  # 45 degrees in radians

print("Sine:", math.sin(angle))
print("Cosine:", math.cos(angle))
print("Tangent:", math.tan(angle))
````
