### Background

A ball that bounces off a wall obeys the **law of reflection**: the angle at which it leaves the wall equals the angle at which it arrived. For a horizontal wall, a ball traveling at an angle of 30 degrees upwards leaves at 30 degrees downwards: its angle changes sign. To move a ball in a direction, a program splits the direction into a horizontal and a vertical part with cosine and sine: at angle *a* and distance *d*, the ball moves *d* × cos *a* to the side and *d* × sin *a* up or down.

This physics is the basis of one of the first video games. In 1958, the physicist William Higinbotham built *Tennis for Two* for a visitors' day at Brookhaven National Laboratory: two players hit a ball over a net, shown on the screen of an oscilloscope, and an analog computer calculated its path [@bnl-tennis]. In this project, you write the rules of a two-player paddle game in which the ball's direction is an angle in degrees, and the `math` module turns it into movement.

### The code you are given

The ball is a dictionary with its position `"x"` and `"y"` in pixels, its direction `"angle"` in degrees and its `"speed"` in pixels per second. As on a screen, `y` grows downwards, so a ball that moves up has a decreasing `y`. The angle is measured counterclockwise from "straight to the right": 0 is right, 90 is up, 180 is left. Two functions are provided. `new_ball()` places a ball in the center, and `bounce_paddle()` reflects it from a paddle, which is a vertical surface, so its angle becomes 180 minus the angle:

```python
def bounce_paddle(ball, x):
    """Reflect the ball from a vertical paddle at position x: the angle becomes 180 minus the angle."""
    ball["x"] = x
    ball["angle"] = (180 - ball["angle"]) % 360


ball = {"x": 470, "y": 160, "angle": 30, "speed": 260}
bounce_paddle(ball, 448)
print(ball["angle"])
```

**Output**

```text
150
```

A ball that flew up and to the right at 30 degrees now flies up and to the left at 150 degrees. The `% 360` keeps the angle between 0 and 359.
