# Scope and Decorators: listing 1

From *Python Programming from First Principles*, Python Basics. [Read this section](/books/001-python-programming-from-first-principles/function-scope-and-decorators.html).

Predict the output, then run the listing.

````python
message = "Hello from the global scope"

def greet():
    local_message = "Hello from inside the function"
    print(local_message)    # accessible only inside the function
    print(message)          # accessing the global variable

greet()
print(message)  # accessible anywhere
````
