```python
# Earlier code from this section

message = "Hello from the global scope"

def greet():
    local_message = "Hello from inside the function"
    print(local_message)    # accessible only inside the function
    print(message)          # accessing the global variable

greet()
print(message)  # accessible anywhere

count = 0

def increase_count():
    global count
    count += 1
    print("Inside function:", count)

increase_count()
print("Outside function:", count)
```
