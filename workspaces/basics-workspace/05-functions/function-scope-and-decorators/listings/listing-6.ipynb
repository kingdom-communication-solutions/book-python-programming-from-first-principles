````python
# Earlier code from this section

message = "Hello from the global scope"

def greet():
    local_message = "Hello from inside the function"
    print(local_message)    # accessible only inside the function
    print(message)          # accessing the global variable

greet()
print(message)  # accessible anywhere

count = 0

def increase_count():
    global count
    count += 1
    print("Inside function:", count)

increase_count()
print("Outside function:", count)

def outer():
    x = "outer value"

    def inner():
        nonlocal x
        x = "inner value"
        print("Inside inner:", x)

    inner()
    print("Inside outer:", x)

outer()
````

Predict the output, then run the listing.

````python
def shout(text):
    return text.upper() + "!"

speak = shout                   # a second name, no call
print(speak("hello"))

def apply_twice(func, value):
    return func(func(value))    # call the function received

print(apply_twice(shout, "hi"))
````
