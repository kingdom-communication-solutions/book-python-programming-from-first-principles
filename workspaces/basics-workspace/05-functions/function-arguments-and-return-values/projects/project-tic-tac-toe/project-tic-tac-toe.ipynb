### How the Game Works

Tic-Tac-Toe, also called noughts and crosses, is a game for two players, X and O, on a board of three rows and three columns. X starts, and the players take turns: in each turn, a player puts their mark in one empty cell. A player who gets three marks in a line, a row, a column or one of the two diagonals, wins, and the other player loses. If all nine cells are full and nobody has three in a line, the game is a draw.

![Three Tic-Tac-Toe boards. First: X starts in the center, O answers in the top-left corner and X plays the top-right corner. Second: X has three marks on a diagonal, so X wins and O loses. Third: all nine cells are full and no line has three equal marks, so the game is a draw.](/books/001-python-programming-from-first-principles/assets/function-arguments-and-return-values/how-to-play.svg "Taking turns, winning and a draw in Tic-Tac-Toe")

The game is small enough for a computer to check every possible game, and such an analysis shows that when neither player makes a mistake, the game ends in a draw. That is why Tic-Tac-Toe has a place in the history of artificial intelligence: game playing was one of the field's first research areas, after Claude Shannon described a chess program in 1950 [@shannon1950; @mccarthy2006], and textbooks still use Tic-Tac-Toe as their first example of a **game tree**, the branching set of all possible moves [@russell2020, chap. 5].

Your rules make the program a referee: they decide which moves are allowed and when a round ends, while choosing a good move is left to the players. Code that is provided for you draws the board and handles the clicks.

### The Board as Nested Lists

A program needs the same board as data. The board has rows, and each row has cells, so a list of lists from [Nested Data Structures](/books/001-python-programming-from-first-principles/collection-nested-data-structures.html) fits it exactly: the board is a list of three rows, and each row is a list of three cells. A cell holds `"X"`, `"O"` or the empty string `""`.

![Left: a three-by-three grid with rows 0 to 2 and columns 0 to 2, row 1 highlighted. Right: the same board as a list of three inner lists, with row 1 highlighted. board[1] is row 1, and board[1][1] is its cell in column 1, which holds X.](diagrams/board-as-nested-list.svg "The board as a list of three rows")

The two indexes follow the board itself: `board[row]` selects a row, and `board[row][col]` one cell in that row. This keeps each rule of the game short:

- Placing a mark is one assignment: `board[row][col] = player`, where the variable `player` holds the mark of the player whose turn it is, `"X"` or `"O"`.
- Checking whether the board is full means visiting every cell with two nested `for` loops, one over the rows and one over the cells of each row.
- Checking a line means comparing three cells, such as `board[0][col]`, `board[1][col]` and `board[2][col]` for a column.

### The rules you are given

```python
def new_board():
    """Return an empty board."""
    return [["", "", ""], ["", "", ""], ["", "", ""]]


def place(board, row, col, player):
    """Put the player's mark in an empty cell and return True; return False if the cell is taken."""
    if board[row][col] != "":
        return False
    board[row][col] = player
    return True


board = new_board()
print(place(board, 1, 1, "X"))
print(place(board, 1, 1, "O"))
print(board[1])
```

**Output**

```text
True
False
['', 'X', '']
```

The text in triple quotes below each `def` line is a **docstring**: a short description of what the function does. `place()` returns `False` when the cell is already taken, so the code that calls it can ignore the click.
