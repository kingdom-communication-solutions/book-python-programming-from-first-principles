```python
# Earlier code from this section

square = lambda x: x ** 2
print(square(5))

def square(x):
    return x ** 2

numbers = [3, 6, 9, 2, 5]
doubled = list(map(lambda x: x * 2, numbers))
print(doubled)

numbers = [3, 6, 9, 2, 5]
squares = [x ** 2 for x in numbers]
evens = [x for x in numbers if x % 2 == 0]
print(squares)
print(evens)

squares = []
for x in numbers:
    squares.append(x ** 2)
print(squares)

words = ["fig", "apple", "banana"]
lengths = {word: len(word) for word in words}
print(lengths)

def factorial(n):
    if n <= 1:
        return 1
    else:
        return n * factorial(n - 1)

print(factorial(5))

def factorial(n):
    if n <= 1:
        return 1
    else:
        return n * factorial(n - 1)

def count_up_to(n):
    count = 1
    while count <= n:
        yield count
        count += 1

# Create a generator
numbers = count_up_to(5)

# Iterate through values one by one
for num in numbers:
    print(num)

gen = count_up_to(3)
print(next(gen))
print(next(gen))
print(next(gen))
```
