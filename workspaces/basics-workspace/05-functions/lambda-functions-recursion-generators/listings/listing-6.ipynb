````python
# Earlier code from this section

square = lambda x: x ** 2
print(square(5))

def square(x):
    return x ** 2

numbers = [3, 6, 9, 2, 5]
doubled = list(map(lambda x: x * 2, numbers))
print(doubled)

numbers = [3, 6, 9, 2, 5]
squares = [x ** 2 for x in numbers]
evens = [x for x in numbers if x % 2 == 0]
print(squares)
print(evens)

squares = []
for x in numbers:
    squares.append(x ** 2)
print(squares)
````
