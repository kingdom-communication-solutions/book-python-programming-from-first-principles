````python
# Earlier code from this section

def add_numbers(*args):
    print("Received:", args)
    print("Sum:", sum(args))

add_numbers(2, 4, 6)
add_numbers(5, 10, 15, 20)

def describe_person(**kwargs):
    print("Person details:")
    for key, value in kwargs.items():
        print(f"{key}: {value}")

describe_person(name="Alice", age=28, city="Paris")

describe_person(name="Bob", job="Engineer")

def show_info(*args, **kwargs):
    print("Positional arguments:", args)
    print("Keyword arguments:", kwargs)
````

Predict the output, then run the listing.

````python
show_info("Python", "Data Science", version=3.12, difficulty="Intermediate")
````
