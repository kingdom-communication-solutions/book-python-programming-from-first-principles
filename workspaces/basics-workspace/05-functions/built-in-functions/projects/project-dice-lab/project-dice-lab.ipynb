### Background

Games with dice were played thousands of years ago, but the mathematics of chance began only in the sixteenth and seventeenth centuries, with questions about dice and card games. Gerolamo Cardano wrote one of the first books on games of chance; it was published in 1663, almost ninety years after his death. In 1654, Blaise Pascal and Pierre de Fermat exchanged letters about how to divide the stakes of a game that had been interrupted, a correspondence often regarded as the beginning of probability theory [@devlin2008]. In 1713, Jacob Bernoulli's *Ars Conjectandi* proved an early form of the **law of large numbers**: the more often an experiment is repeated, the closer the observed frequencies come to the underlying probabilities [@hald1990].

A computer can repeat an experiment thousands of times in a second. Such a **simulation** is one of the everyday tools of data science: when a probability is hard to calculate, you can estimate it by running the experiment many times and counting the results. A fair die shows each face with probability 1/6, so the average of many rolls approaches (1 + 2 + 3 + 4 + 5 + 6) / 6 = 3.5. In this project, you roll a die with the `random` module, count a thousand rolls in a dictionary and watch the counts even out.

### The code you are given

The counts are stored in a dictionary that maps each face to the number of times it was rolled. The function `average()` is provided; it calculates the mean face from such a dictionary:

```python
def average(counts):
    """Return the mean face of the rolls in counts, a dictionary {face: number of rolls}."""
    rolls = sum(counts.values())
    if rolls == 0:
        return 0
    total = 0
    for face, number in counts.items():
        total += face * number
    return total / rolls


print(average({1: 2, 2: 0, 3: 0, 4: 0, 5: 0, 6: 2}))
```

**Output**

```text
3.5
```

Two ones and two sixes give (1 + 1 + 6 + 6) / 4 = 3.5. The test `rolls == 0` prevents a division by zero before anything has been rolled.
