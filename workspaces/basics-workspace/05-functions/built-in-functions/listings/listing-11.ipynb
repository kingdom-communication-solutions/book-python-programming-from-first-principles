```python
# Earlier code from this section

print(list(range(5)))        # 0 to 4
print(list(range(2, 8)))     # 2 to 7
print(list(range(0, 10, 2))) # even numbers

numbers = [3, 1, 4, 2]
print(sorted(numbers))               # ascending order
print(sorted(numbers, reverse=True)) # descending order

words = ["banana", "fig", "apple"]
print(sorted(words, key=len))

def square(x):
    return x ** 2

numbers = [1, 2, 3, 4]
result = map(square, numbers)
print(list(result))

def is_even(x):
    return x % 2 == 0

numbers = [1, 2, 3, 4, 5, 6]
even_numbers = filter(is_even, numbers)
print(list(even_numbers))

import random

print(random.random())

print(random.randint(1, 6))  # simulating a dice roll

fruits = ["apple", "banana", "cherry", "date"]
print(random.choice(fruits))
```
