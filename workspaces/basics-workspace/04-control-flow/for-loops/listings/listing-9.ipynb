```python
# Earlier code from this section

fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print("I like", fruit)

for letter in "Python":
    print(letter)

for i in range(5):
    print("Iteration", i)

for i in range(2, 10, 2):
    print(i)

readings = [12, 7, -3, 9, -8]

for value in readings:
    if value < 0:
        print("First negative reading:", value)
        break
    print("Checked", value)

for row in range(1, 4):
    line = ""
    for column in range(1, 6):
        line += f"{row * column} "
    print(line)

seats = [
    ["free", "taken", "free"],
    ["taken", "taken", "free"],
]

free = 0
for row in seats:
    for seat in row:
        if seat == "free":
            free += 1

print("Free seats:", free)
```
