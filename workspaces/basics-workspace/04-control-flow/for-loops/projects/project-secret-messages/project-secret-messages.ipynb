### Background

Hiding a message by replacing its letters is one of the oldest known uses of a fixed procedure. The Roman historian Suetonius reports that Julius Caesar wrote confidential letters by replacing each letter with the one three places further on in the alphabet, so that D stands for A [@suetonius1913, Julius 56.6]. Such a **shift cipher** is easy to apply and, for a long time, was hard to read for anyone without the key. In the ninth century, the Arab scholar al-Kindi described **frequency analysis**: in any language, some letters are much more common than others, so the most frequent letters of a long secret message reveal the substitution [@alkadi1992]. A shift cipher is weaker still. It has only 26 possible keys, so a computer can try them all in an instant. Modern cryptography starts from the same question as Caesar and al-Kindi: how hard is it to read a message without its key?

In this project, a `for` loop works through a message one character at a time. Each letter is found in the alphabet with `find()`, moved a number of places, and wrapped from `z` back to `a` with the remainder operator `%`.

The project is a mission rather than a game: you agree on a shift with a partner, send a message only the two of you are meant to read, and recover the messages you receive. The program is complete when it turns a lowercase message into its secret form and back again, leaves spaces and punctuation unchanged, and, in the extension, finds the shift of an intercepted message. Along the way, you will see why such a cipher protects nothing.

```{.python .intentional-error}
alphabet = "abcdefghijklmnopqrstuvwxyz"
message = "meet me at the zoo at noon"
shift = 3

secret = ""
for letter in message:
    position = alphabet.find(letter)
    if position == -1:      # a space or punctuation mark stays as it is
        secret += letter
    else:
        secret += alphabet[position + shift]

print("Secret:", secret)
```
