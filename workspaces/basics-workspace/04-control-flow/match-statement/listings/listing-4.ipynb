# The `match` Statement: listing 4

From *Python Programming from First Principles*, Python Basics. [Read this section](/books/001-python-programming-from-first-principles/match-statement.html).

Predict the output, then run the listing.

````python
choice = int(input("Enter a menu option (1–3): "))

match choice:
    case 1:
        print("You selected: Start new game")
    case 2:
        print("You selected: Load game")
    case 3:
        print("You selected: Exit")
    case _:
        print("Invalid option.")
````
