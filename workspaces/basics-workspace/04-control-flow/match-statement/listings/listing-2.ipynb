# The `match` Statement: listing 2

From *Python Programming from First Principles*, Python Basics. [Read this section](/books/001-python-programming-from-first-principles/match-statement.html).

Predict the output, then run the listing.

```python
day = input("Enter a day of the week: ")

match day:
    case "Monday":
        print("Start of the work week!")
    case "Friday":
        print("Almost the weekend!")
    case "Saturday" | "Sunday":
        print("It’s the weekend, enjoy your rest!")
    case _:
        print("Midweek day — keep going!")
```
