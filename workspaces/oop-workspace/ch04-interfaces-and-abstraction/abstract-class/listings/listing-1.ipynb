# Abstract Classes: listing 1

From *Python Programming from First Principles*, Chapter 7. [Read this section](/books/001-python-programming-from-first-principles/abstract-class.html).

Predict the output, then run the listing.

````python
from abc import ABC, abstractmethod


class Notifier(ABC):
    @abstractmethod
    def send(self, member, message):
        """Deliver one message to a member."""


class ConsoleNotifier(Notifier):
    def send(self, member, message):
        print(f"To {member.name}: {message}")


notifier = ConsoleNotifier()
print(isinstance(notifier, Notifier))
````
