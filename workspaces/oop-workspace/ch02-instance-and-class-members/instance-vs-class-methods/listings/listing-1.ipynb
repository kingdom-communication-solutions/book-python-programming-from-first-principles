# Instance and Class Methods: listing 1

From *Python Programming from First Principles*, Chapter 7. [Read this section](/books/001-python-programming-from-first-principles/instance-vs-class-methods.html).

Predict the output, then run the listing.

````python
class Book:
    loan_days = 14

    def __init__(self, copy_id, title, author, year):
        self.copy_id = copy_id
        self.title = title
        self.author = author
        self.year = year

    @classmethod
    def from_record(cls, record):
        copy_id, title, author, year = record.split(";")
        return cls(copy_id, title, author, int(year))

    def description(self):
        return f"{self.title} by {self.author} ({self.year})"

city = Book.from_record("B003;City Letters;K. Owusu;2019")
print(city.copy_id, city.description())
````
