````python
class Book:
    loan_days = 14

    def __init__(self, copy_id, title, author, year):
        self.copy_id = copy_id
        self.title = title
        self.author = author
        self.year = year

    @classmethod
    def from_record(cls, record):
        copy_id, title, author, year = record.split(";")
        return cls(copy_id, title, author, int(year))

    @classmethod
    def set_loan_days(cls, days):
        cls.loan_days = days

    @staticmethod
    def is_valid_copy_id(copy_id):
        return len(copy_id) == 4 and copy_id[0] == "B" and copy_id[1:].isdigit()

    def description(self):
        return f"{self.title} by {self.author} ({self.year})"

city = Book.from_record("B003;City Letters;K. Owusu;2019")
Book.set_loan_days(21)
print(Book.loan_days, city.loan_days)
print(Book.is_valid_copy_id("B004"))
print(Book.is_valid_copy_id("4"))
print(city.is_valid_copy_id("B12"))
````
