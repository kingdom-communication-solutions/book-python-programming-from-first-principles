````python
# Earlier code from this section

class Book:
    loan_days = 14

    def __init__(self, copy_id, title):
        self.copy_id = copy_id
        self.title = title

river = Book("B001", "River Stories")
maps = Book("B002", "Silent Maps")
print(Book.loan_days, river.loan_days, maps.loan_days)

Book.loan_days = 21
print(river.loan_days, maps.loan_days)

river.loan_days = 7
print(river.loan_days, maps.loan_days, Book.loan_days)
````
