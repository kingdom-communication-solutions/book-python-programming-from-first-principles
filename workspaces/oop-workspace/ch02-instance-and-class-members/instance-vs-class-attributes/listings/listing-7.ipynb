```python
# Earlier code from this section

class Book:
    loan_days = 14

    def __init__(self, copy_id, title):
        self.copy_id = copy_id
        self.title = title

river = Book("B001", "River Stories")
maps = Book("B002", "Silent Maps")
print(Book.loan_days, river.loan_days, maps.loan_days)

Book.loan_days = 21
print(river.loan_days, maps.loan_days)

river.loan_days = 7
print(river.loan_days, maps.loan_days, Book.loan_days)

class Book:
    created_count = 0

    def __init__(self, copy_id, title):
        self.copy_id = copy_id
        self.title = title
        Book.created_count += 1

Book("B001", "River Stories")
Book("B002", "Silent Maps")
print(Book.created_count)

class Book:
    tags = []

    def __init__(self, title):
        self.title = title

river = Book("River Stories")
maps = Book("Silent Maps")
river.tags.append("history")
print(maps.tags)
print(river.tags is maps.tags)

class Book:
    def __init__(self, title):
        self.title = title
        self.tags = []

river = Book("River Stories")
maps = Book("Silent Maps")
river.tags.append("history")
print(river.tags, maps.tags)
```
