````python
# Earlier code from this section

class Book:
    pass

river = Book()
river.title = "River Stories"
river.author = "A. Mensah"
river.year = 2024

print(river.title)
print(f"{river.title} by {river.author} ({river.year})")

class Book:
    def description(self):
        return f"{self.title} by {self.author} ({self.year})"

river = Book()
river.title = "River Stories"
river.author = "A. Mensah"
river.year = 2024

print(river.description())

print(Book.description(river))

print(river.description)
````

````python
class Temperature:
    def to_fahrenheit(self):
        return self.celsius * 9 / 5 + 32

t1 = Temperature()
t1.celsius = 25
t2 = Temperature()
t2.celsius = 49

print(f"{t1.celsius}°C is {t1.to_fahrenheit()}°F")
print(f"{t2.celsius}°C is {t2.to_fahrenheit()}°F")
````
