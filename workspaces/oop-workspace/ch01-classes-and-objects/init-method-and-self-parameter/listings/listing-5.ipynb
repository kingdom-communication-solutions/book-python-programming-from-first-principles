# Initializing Objects with __init__ and self: listing 5

From *Python Programming from First Principles*, Chapter 7. [Read this section](/books/001-python-programming-from-first-principles/init-method-and-self-parameter.html).

Run the first cell once. It repeats the earlier code of this section that the listing builds on, so it may print earlier results.

```python
# Earlier code from this section

class Book:
    def __init__(self, copy_id, title, author, year):
        self.copy_id = copy_id
        self.title = title
        self.author = author
        self.year = year

    def description(self):
        return f"{self.title} by {self.author} ({self.year})"

river = Book("B001", "River Stories", "A. Mensah", 2024)
print(river.copy_id)
print(river.description())

class Book:
    def __init__(self, copy_id, title, author, year, language="English"):
        self.copy_id = copy_id
        self.title = title
        self.author = author
        self.year = year
        self.language = language

first = Book("B001", "River Stories", "A. Mensah", 2024)
translation = Book("B005", "Contes du fleuve", "A. Mensah", 2025, "French")
print(first.language, translation.language)
```
