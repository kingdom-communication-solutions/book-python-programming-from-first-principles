# Encapsulation: listing 1

From *Python Programming from First Principles*, Chapter 7. [Read this section](/books/001-python-programming-from-first-principles/encapsulation.html).

Predict the output, then run the listing.

````python
class Book:
    def __init__(self, copy_id, title):
        self.copy_id = copy_id
        self.title = title
        self._available = True

    def is_available(self):
        return self._available

    def check_out(self):
        if not self._available:
            raise ValueError(f"{self.copy_id} is already on loan")
        self._available = False

    def check_in(self):
        if self._available:
            raise ValueError(f"{self.copy_id} is not on loan")
        self._available = True

river = Book("B001", "River Stories")
river.check_out()
print(river.is_available())

try:
    river.check_out()
except ValueError as error:
    print("Refused:", error)
````
