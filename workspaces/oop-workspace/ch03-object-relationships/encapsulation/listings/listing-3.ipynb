Run the first cell once. It repeats the earlier code of this section that the listing builds on, so it may print earlier results.

```python
# Earlier code from this section

class Book:
    def __init__(self, copy_id, title):
        self.copy_id = copy_id
        self.title = title
        self._available = True

    def is_available(self):
        return self._available

    def check_out(self):
        if not self._available:
            raise ValueError(f"{self.copy_id} is already on loan")
        self._available = False

    def check_in(self):
        if self._available:
            raise ValueError(f"{self.copy_id} is not on loan")
        self._available = True

river = Book("B001", "River Stories")
river.check_out()
print(river.is_available())

try:
    river.check_out()
except ValueError as error:
    print("Refused:", error)

class Book:
    def __init__(self, copy_id, title):
        self.copy_id = copy_id
        self.title = title
        self._available = True

    @property
    def available(self):
        return self._available

    def check_out(self):
        if not self._available:
            raise ValueError(f"{self.copy_id} is already on loan")
        self._available = False

river = Book("B001", "River Stories")
print(river.available)
river.check_out()
print(river.available)
```
