````python
# An assistant's answer: it contains one mistake.
class Book:
    def __init__(self, title):
        self.title = title


class Library(Book):
    def __init__(self, name):
        super().__init__(name)
        self.books = []

    def add(self, book):
        self.books.append(book)


central = Library("Central")
central.add(Book("River Stories"))
central.add(central)
print(central.title, len(central.books))
````
