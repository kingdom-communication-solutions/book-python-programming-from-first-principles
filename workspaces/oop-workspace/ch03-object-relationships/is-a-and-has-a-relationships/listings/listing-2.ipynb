```python
# Earlier code from this section

class Book:
    def __init__(self, copy_id, title):
        self.copy_id = copy_id
        self.title = title


class ListCatalogue(list):
    def add(self, book):
        if any(item.copy_id == book.copy_id for item in self):
            raise ValueError(f"Duplicate copy ID: {book.copy_id}")
        self.append(book)


catalogue = ListCatalogue()
catalogue.add(Book("B001", "River Stories"))
try:
    catalogue.add(Book("B001", "River Stories"))
except ValueError as error:
    print("Refused:", error)

catalogue.append(Book("B001", "River Stories"))
catalogue.insert(0, "not a book")
print(len(catalogue))
```

```python
# An assistant's answer: it contains one mistake.
class Book:
    def __init__(self, title):
        self.title = title


class Library(Book):
    def __init__(self, name):
        super().__init__(name)
        self.books = []

    def add(self, book):
        self.books.append(book)


central = Library("Central")
central.add(Book("River Stories"))
central.add(central)
print(central.title, len(central.books))
```
