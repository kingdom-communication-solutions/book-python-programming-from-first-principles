# “Is-a” and “Has-a” Relationships: listing 1

From *Python Programming from First Principles*, Chapter 7. [Read this section](/books/001-python-programming-from-first-principles/is-a-and-has-a-relationships.html).

Predict the output, then run the listing.

```python
class Book:
    def __init__(self, copy_id, title):
        self.copy_id = copy_id
        self.title = title


class ListCatalogue(list):
    def add(self, book):
        if any(item.copy_id == book.copy_id for item in self):
            raise ValueError(f"Duplicate copy ID: {book.copy_id}")
        self.append(book)


catalogue = ListCatalogue()
catalogue.add(Book("B001", "River Stories"))
try:
    catalogue.add(Book("B001", "River Stories"))
except ValueError as error:
    print("Refused:", error)

catalogue.append(Book("B001", "River Stories"))
catalogue.insert(0, "not a book")
print(len(catalogue))
```
